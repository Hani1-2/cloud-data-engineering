```python

import pandas as pd
from kafka import KafkaProducer
from time import sleep
from json import dumps
import json
```

```python

producer = KafkaProducer(bootstrap_servers=['54.87.131.217:9092'], #change ip here
                         value_serializer=lambda x: 
                         dumps(x).encode('utf-8'))
```

```python
import socket
sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
result = sock.connect_ex(('54.87.131.217', 9092))
print('Port is open' if result == 0 else f'Connection failed: {result}')
sock.close()
```

```python

producer.send('demo_test', value={'realtime':'is fun'})
```

```python

df = pd.read_csv("indexProcessed.csv")
```

```python

while True:
    dict_stock = df.sample(1).to_dict(orient="records")[0]
    producer.send('demo_test', value=dict_stock)
    sleep(1)
```

```python
for i in range(18):
    dict_stock = df.sample(15).to_dict(orient="records")[0]
    producer.send('demo_test', value=dict_stock)
    sleep(1)
```
